```python
# K Means Cluster creation
from sklearn.cluster import KMeans
```

```python
kmeans =KMeans(n_clusters=2)
```

```python
#Fit the model to all data except for the Private lable
kmeans.fit(df.drop('Private',axis=1))
```

```python
kmeans.cluster_centers_
```

```python
kmeans.labels_
```

```python
#Evaulation 
#df.head()
#Private---is a categorical columns need to convert into numerical columns
#Createing a new column for df called 'Cluster',Which is a 1 for a private school,a dn a o for a public school
df['Cluster'] = df['Private'].map({'Yes': 1, 'No': 0}) # easy code
#Another way
#df['Cluster'] = df['Private'].apply(lambda x: 1 if x == 'Yes' else 0)
#Another way 
#def converter(private):
#    if private=='Yes':
#        return 1
#    else:
#        return 0

#df['Cluster']=df['private'].apply(converter)
```

```python
pred_labels = kmeans.labels_
```

```python
#Create a confusion matrix and classification report to see how eell the Kmeans claustering workded without being given anyalabels
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(confusion_matrix(df['Cluster'],pred_labels))
print('\n')
print(classification_report(df['Cluster'],pred_labels))
```
