```python
#NLP classification Task
#our task is to grab reviews that wer either  star or 5 stars
#Create a dataframe called yelp_class that contains the columns of yelp dataframe bit for only the  or 5 stars review.
```

```python
df_class = df[(df['stars']==1) | (df['stars']==5)]
```

```python
df_class.info()
```

```python
#Create two objects X and y,
X =df_class['text']
y=df_class['stars']
```

```python
#import CountVectorixer and create a countVectorizer object
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer()
```

```python
X = cv.fit_transform(X)
```

```python
#Train test Split
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
#Training a Model
#Importing MultinomialNB and create an instance of the estimator and call is nb. # We can use Random forest classifier
```

```python
from sklearn.naive_bayes import MultinomialNB
nb = MultinomialNB()
```

```python
#Now fit nb using the traning data
nb.fit(X_train,y_train)
```

```python
#Predictions and Evalautions
predictions = nb.predict(X_test)
```

```python
#create a confusion matrix and classification report using these predcitions and y_test
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(confusion_matrix(y_test,predictions))
print('\n')
print(classification_report(y_test,predictions))
```

```python
#Using Text Processing
```

```python
# importing TidfTransform from sklearn
from sklearn.feature_extraction.text import TfidfTransformer
```

```python
#Importing pipeline from sklearn
from sklearn.pipeline import Pipeline
```

```python
#Create a pipeline with the following steps: CountVectorizer(),TfidfTransformer(),MuiltinomialNB(0
```

```python
pipeline = Pipeline([
    ('bow', CountVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('model', MultinomialNB())  # Corrected tuple and parenthesis
])
```

```python
#using The Pipeline
# #we need ot rewrite the original data (remeber that we overwrote X as the CountVectorized version .What we need is just the tes)
```

```python
#Train Test Split
#Redo the train test split on the df_class object
X= df_class['text']
y = df_class['stars']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
pipeline.fit(X_train,y_train)
```

```python
#Prediction and Evaulautaion
predictions = pipeline.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(confusion_matrix(y_test,predictions))
print('\n')
print(classification_report(y_test,predictions))
```
